```python
# --- SISTEMA IA: GESTIÓN DE EMERGENCIAS PARA BOMBEROS ---

# Al inicio la lista de reportes empieza vacía
reportes_emergencia = []

# Variable para controlar la ejecución del programa
ejecutar_sistema = "si"

while ejecutar_sistema == "si":
    print("\n========================================")
    print("   CENTRO DE MANDO INTELIGENTE (IA)   ")
    print("========================================")
    print("1. Registrar nueva emergencia")
    print("2. Ver lista de reportes (Historial)")
    print("3. IA: Realizar diagnóstico de prioridad")
    print("4. Salir del sistema")

    opcion = input("\nSeleccione una opción (1-4): ")

    # OPCIÓN 1: Registrar datos en la lista
    if opcion == "1":
        print("\n[REGISTRO] Ingrese datos del incidente:")
        ubicacion = input("Ubicación del incendio: ")
        temperatura = float(input("Temperatura detectada (Celsius): "))
        victimas = input("¿Hay presencia de víctimas? (si/no): ").lower()

        # Guardamos los datos en una sublista y la agregamos a la lista principal
        nuevo_reporte = [ubicacion, temperatura, victimas]
        reportes_emergencia.append(nuevo_reporte)
        print("¡Emergencia registrada con éxito!")

    # OPCIÓN 2: Mostrar la lista usando ciclo WHILE (en lugar de for)
    elif opcion == "2":
        print("\n--- HISTORIAL DE EMERGENCIAS ---")
        if len(reportes_emergencia) == 0:
            print("No hay reportes en la base de datos.")
        else:
            # Encabezado perfectamente alineado en columnas fijas
            print(f"{'ID':<5} | {'Ubicación del Incendio':<25} | {'Temperatura':<15} | {'Víctimas':<10}")
            print("-" * 68) # Línea divisoria decorativa

            indice = 0
            # Usamos while para recorrer la lista sin usar el ciclo for
            while indice < len(reportes_emergencia):
                datos = reportes_emergencia[indice]

                # Formateamos las variables para que tengan un ancho fijo y no se muevan las columnas
                id_str = f"{indice}"
                lugar_str = datos[0]
                temp_str = f"{datos[1]}°C"
                victimas_str = datos[2].upper() # Lo ponemos en mayúsculas para que resalte

                # El operador :< define el tamaño de cada casilla visual
                print(f"{id_str:<5} | {lugar_str:<25} | {temp_str:<15} | {victimas_str:<10}")
                indice = indice + 1

    # OPCIÓN 3: Inteligencia Artificial (Lógica de decisión)
    elif opcion == "3":
        if len(reportes_emergencia) == 0:
            print("\n[IA] Error: No hay datos para analizar.")
        else:
            id_analizar = int(input("\nIngrese el ID del reporte para diagnóstico IA: "))

            # Verificamos si el ID es válido con operadores relacionales
            if id_analizar < 0 or id_analizar >= len(reportes_emergencia):
                print("ID no válido.")
            else:
                incidente = reportes_emergencia[id_analizar]
                lugar = incidente[0]
                temp = incidente[1]
                hay_victimas = incidente[2]

                print(f"\n[IA] Analizando incidente en: {lugar}...")

                # Motor de reglas de la IA (if, elif, else y operadores lógicos)
                if temp > 100 and hay_victimas == "si":
                    prioridad = "CRÍTICA: Rescate extremo inmediato."
                elif temp > 100 or hay_victimas == "si":
                    prioridad = "ALTA: Despliegue de equipo pesado."
                elif temp > 50 and temp <= 100:
                    prioridad = "MEDIA: Asegurar el perímetro."
                else:
                    prioridad = "BAJA: Incendio controlado o bajo vigilancia."

                print(f"DIAGNÓSTICO FINAL: {prioridad}")

    # OPCIÓN 4: Salir
    elif opcion == "4":
        print("\nCerrando sistema de IA... ¡Manténganse a salvo, bombero!")
        print("Programa realizado por Anais Medina NL = 40")
        ejecutar_sistema = "no"

    # Manejo de opciones no válidas
    else:
        print("\n[Error] Opción no válida. Intente de nuevo.")

print("\n--- Sistema fuera de línea ---")
```
